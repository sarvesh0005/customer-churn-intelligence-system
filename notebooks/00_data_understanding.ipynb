# Business Understanding

## Objective

Predict whether a telecom customer is likely to churn.

## Business Goal

Help the company identify customers at high risk of leaving so proactive retention strategies can be applied.

## Target Variable

Churn

Yes → Customer left

No → Customer stayed

Dataset Name

IBM Telco Customer Churn

Source

IBM Sample Dataset

Domain

Telecommunications

Prediction Type

Binary Classification
